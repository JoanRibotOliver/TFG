# Prominence in the solar corona
Magnetic field lines aren't properly drawn but it gives more or less the idea.

```python
import matplotlib.pyplot as plt
import numpy as np
import scienceplots
plt.style.use('science')

fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(-6, 6)
ax.set_ylim(-2, 6)
ax.axis('off')

# Línia gruixada de la superfície
ax.plot([-6, 6], [0, 0], color='black', linewidth=3)

# Línies diagonals sota la superfície (simulant "terra" o "superfície")
for x in np.linspace(-5.8, 5.8, 75):
    ax.plot([x - 0.1, x + 0.1], [-0.3, -0.1], color='black', linewidth=1)

# Textos
ax.text(0, 5.5, r'$\vec{B} =$ línies de camp', fontsize=14, ha='center')
ax.text(-4, 4.5, 'corona', color='green', fontsize=12)
ax.text(3.5, 4.5, 'corona', color='green', fontsize=12)
ax.text(0.7, 3, 'protuberància', color='blue', fontsize=12)

# Protuberància solar
protuberancia = plt.Rectangle((-0.5, 0), 1.0, 3, color='blue', alpha=0.4)
ax.add_patch(protuberancia)

# Línies de camp magnètic
x = np.linspace(-15, 15, 400)
def field_line(x, height, scale):
    return height * np.exp(-x**4 + x**2 + 0.25 / scale)

heights = [1, 2, 3]
scales = [1.5, 2.5, 4]

for h, s in zip(heights, scales):
    y = field_line(x, h, s)
    ax.plot(x, y, color='black')

# Puntes de fletxa (només la punta) sobre línies de camp
arrow_positions = [(-3.5, 2), (-1.8, 2.5), (1.8, 2.5), (3.5, 2)]
for x0, h in arrow_positions:
    s = 2.5 if abs(x0) < 3 else 4
    y0 = field_line(x0, h, s)
    dy = 0.001
    slope = (field_line(x0 + dy, h, s) - y0) / dy
    angle = np.arctan(slope)
    dx = 0.2 * np.cos(angle)
    dy = 0.2 * np.sin(angle)
    ax.arrow(x0, y0, dx, dy, head_width=0.15, head_length=0.2, fc='black', ec='black', length_includes_head=True)

plt.title("Línies de camp i protuberància solar", fontsize=14)
plt.show()
```

 # Representation of the 2D problem to be solved (F&S modes)


```python
import matplotlib.pyplot as plt
import numpy as np
import scienceplots
plt.style.use('science')
plt.rcParams.update({
    "font.size": 20,               # Mida general del text
    "axes.titlesize": 20,         # Mida del títol dels eixos
    "axes.labelsize": 20,         # Mida de les etiquetes dels eixos
    "xtick.labelsize": 20,        # Mida dels ticks de l'eix X
    "ytick.labelsize": 20,        # Mida dels ticks de l'eix Y
    "legend.fontsize": 20,        # Mida del text de la llegenda
    "figure.dpi": 300,            # Resolució
    "text.usetex": True,          # Activa LaTeX per al text
    "mathtext.fontset": "cm",     # (Opcional) Usa Computer Modern si no s’usa LaTeX
#    #"font.serif": ["Computer Modern"],  # Use Computer Modern (LaTeX default)
#    #"font.family": "serif",    # Optional: match LaTeX font
})


# Configuració del gràfic
plt.figure(figsize=(12, 7))

ax = plt.gca()
ax.set_frame_on(False)
ax.axis('off')


# Definim els límits de la protuberància (amplada reduïda)
L = 1.0
x = np.linspace(-L, L, 100)
z = np.linspace(0, 2, 100)

# Dibuixem la protuberància (rectangle central més petit)
plt.fill_between(x, 0, 2, where=(x >= -0.25) & (x <= 0.25), color='blue', alpha=0.5, label='Protuberància')

# Dibuixem les superfícies laterals (Sol)
plt.fill_betweenx(z, -L-0.1, -L, color='gray', alpha=0.3, label='Superfície')
plt.fill_betweenx(z, L, L+0.1, color='gray', alpha=0.3)

# Eixos i etiquetes
#plt.axhline(0, color='black', linewidth=1)
#plt.axvline(0, 0, 2, color='black', linewidth=1)

# Etiquetes de les regions
plt.text(-L-0.15, 2.05, 'superfície', fontsize=14, color='gray')
plt.text(L-0.05, 2.05, 'superfície', fontsize=14, color='gray')
plt.text(-L+0.3, 1.8, 'corona', fontsize=14, color='green')
plt.text(L-0.5, 1.8, 'corona', fontsize=14, color='green')
plt.text(-0.4, 2.05, 'protuberància', fontsize=14, color='blue')

# Fletxa x
plt.arrow(0, 1, 1.15, 0, head_width=0.04, head_length=0.08, fc='black', ec='black', linewidth=1, zorder=10)
plt.text(1.2, 0.9, r'$x$', fontsize=14, color='black', ha='center')

# Fletxa z
plt.arrow(0, 0, 0, 2.1, head_width=0.04, head_length=0.1, fc='black', ec='black', linewidth=1, zorder=10)
plt.text(0.05, 2.1, r'$z$', fontsize=14, color='black', ha='center')

# Draw the y-axis (coming out of the screen) from (0, 1)
y_arrow_origin = (0, 1)
y_arrow_dx = -0.4
y_arrow_dz = -0.4
plt.arrow(y_arrow_origin[0], y_arrow_origin[1], y_arrow_dx, y_arrow_dz,
          head_width=0.04, head_length=0.1, fc='black', ec='black', linewidth=1, zorder=10)
# Add y-axis label
plt.text(y_arrow_origin[0] + y_arrow_dx - 0.075, y_arrow_origin[1] + y_arrow_dz, r'$y$', fontsize=14, color='black')


# Dibuixem línies horitzontals discontínues
z_values = [0.25, 1.0, 1.75]
for z_val in z_values:
    plt.plot(x, [z_val]*len(x), linestyle='-', color='black', alpha=1, linewidth=1.5)

plt.text(0.5, 1.05, r'$\vec{B_0}$', fontsize=14, color='black')

# Afegim puntes de fletxa a les línies horitzontals
arrow_length = 0.1 
for z_val in z_values:
    plt.arrow(-0.75, z_val, arrow_length, 0, head_width=0.04, head_length=0.1, fc='black', ec='black', length_includes_head=True)
    plt.arrow(0.6 - arrow_length, z_val, arrow_length, 0, head_width=0.05, head_length=0.1, fc='black', ec='black', length_includes_head=True)

# Remove z axis numbers and add H at the top
plt.yticks([])  # Remove y axis numbers
plt.xticks([])  # Remove x axis numbers

# Afegim H a la part superior
plt.text(0.275, 2, r'$H$', fontsize=14, color='black', ha='center')  # Add H at the top center# Add y-axis label

# Etiquetes proportions
plt.text(L+0.031, 0.02, r'$x_c$', fontsize=14)
plt.text(-L-0.085, 0.02, r'$-x_c$', fontsize=14)

plt.text(L/10 + 0.15, 0.02, r'$x_p$', fontsize=14)
plt.text(-L/10 -0.225, 0.02, r'$-x_p$', fontsize=14)


# Configuració final
#plt.title('Representació d\'una Protuberància Solar (2D)')
#plt.xlabel('x')
#plt.ylabel('z')
plt.xlim(-L-0.25, L+0.25)
plt.ylim(0, 2.25)
plt.savefig('protuberancia_solar_2d.png', dpi=300, bbox_inches='tight')
plt.show()
```

# Longitudinal structure 1D string

```python
import matplotlib.pyplot as plt
import scienceplots
plt.style.use('science')

# Longituds
lc = 5
lp = 2
x0 = 0
x1 = x0 + lc
x2 = x1 + lp
x3 = x2 + lc

fig, ax = plt.subplots(figsize=(10, 2))
ax.set_xlim(-1, x3 + 1)
ax.set_ylim(-1, 1)
ax.axis('off')

# Trams de la corda
ax.plot([x0, x1], [0, 0], color='black', linewidth=2)
ax.plot([x1, x2], [0, 0], color='blue', linewidth=4)
ax.plot([x2, x3], [0, 0], color='black', linewidth=2)

# Línies als extrems (negres i gruixudes)
ax.plot([x0, x0], [-0.35, 0.4], color='black', linewidth=3)
ax.plot([x3, x3], [-0.35, 0.4], color='black', linewidth=3)

# Línies de canvi de densitat (grises)
ax.plot([x1, x1], [-0.25, 0.25], color='gray', linewidth=2, alpha=0.5)
ax.plot([x2, x2], [-0.25, 0.25], color='gray', linewidth=2, alpha=0.5)

# Etiquetes de densitat
ax.text((x0 + x1)/2, 0.2, r'$\rho_c$', fontsize=16, color='green', ha='center')
ax.text((x1 + x2)/2, 0.3, r'$\rho_p$', fontsize=16, color='blue', ha='center')
ax.text((x2 + x3)/2, 0.2, r'$\rho_c$', fontsize=16, color='green', ha='center')


# Etiquetes de longitud
ax.text((x1 + x2)/2 - 1, -0.6, r'$-x_p$', fontsize=14, color='black', ha='center')
ax.text((x1 + x2)/2 + 1, -0.6, r'$x_p$', fontsize=14, color='black', ha='center')
ax.text(0, -0.6, r'$-L$', fontsize=14, color='black', ha='center')
ax.text(12, -0.6, r'$L$', fontsize=14, color='black', ha='center')



# Suports ratllats
for x_support in [x0 , x3 + 0.2]:
    for y in range(-5, 6):
        ax.plot([x_support - 0.2, x_support], [y * 0.05, y * 0.05 + 0.1], color='black')

# Títol opcional
plt.title("Corda 1D amb densitat variable", fontsize=14)
plt.tight_layout()
plt.show()
```
