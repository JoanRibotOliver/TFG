# Intent fer un contador de paraules per saber quines són candidates a ser "paraules claus"

```python
import re
from collections import Counter

with open("Cos.tex", encoding="utf-8") as f:
    text = f.read()

# Remove LaTeX commands and braces
text = re.sub(r"\\[a-zA-Z]+\*?(?:\[[^\]]*\])?(?:\{[^}]*\})*", " ", text)
text = re.sub(r"[%$&{}]", " ", text)
# Remove numbers and punctuation, lowercase everything
words = re.findall(r"\b\w+\b", text.lower())
# Filter out very short words (optional, e.g., length > 2)
words = [w for w in words if len(w) > 2]
counter = Counter(words)
for word, count in counter.most_common(50):
    print(f"{word}: {count}")
```
